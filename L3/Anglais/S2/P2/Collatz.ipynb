This article from Scientific American deals with the Collatz conjecture (or Syracuse problem, or 3n+1 conjecture, as it is most commonly known), a mathematical problem that involves determining whether, for any natural number, the sequence obtained by applying the following rules always converges to 1: if the number is odd, multiply it by 3 and add 1; if it is even, divide it by 2. Despite the simplicity of these rules, no mathematician has yet succeeded in proving that the sequence always converges to 1, or conversely, that it can form another cycle or diverge.

The article highlights mathematicians' enthusiasm for this conjecture, with anecdotes about entire universities grappling with the problem unsuccessfully. It also explores the history of the conjecture, from its uncertain origin to its dissemination in mathematical circles in the 1950s and 1960s.

A portion of the article examines empirical evidence for the conjecture, including computer projects that have verified the validity of the conjecture for considerable numbers, thereby reinforcing the idea that the conjecture could be true for all natural numbers.

However, the article does not fail to mention the potential limits of the conjecture. It discusses the possibility of non-trivial loops that may not converge to 1, although their existence is unlikely. It also mentions the work of Terence Tao, who showed that almost all natural numbers eventually reach 1, but without providing a general proof.

Finally, the article explores the possibility that the Collatz conjecture may remain unresolved indefinitely. It brings up John Horton Conway's reflections on the intrinsically unsolvable nature of some mathematical problems, suggesting that the Collatz conjecture might be one of these cases.

In summary, this Scientific American article offers a comprehensive overview of the Collatz conjecture, from its history to its current implications, highlighting both the allure and the complexity of this deceptively simple mathematical problem.
