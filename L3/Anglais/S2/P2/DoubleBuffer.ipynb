**Intention**
Double buffering is a technique intended to make a series of sequential operations instantaneous or simultaneous, creating an illusion of fluidity for the user.

**Motivation**
Computers execute tasks sequentially. However, for applications like video games, it is crucial that users perceive actions as occurring instantly. Although multi-threaded and multi-core architectures allow for some concurrency, they do not completely solve this problem.

**How Computer Graphics Work**
A computer screen displays pixels sequentially, line by line, very quickly (about 60 times per second), which creates for the human eye the illusion of a continuous, stable image. This image is constructed from data stored in a framebuffer, where each pixel is represented by color values stored in memory.

**Issue**
When rendering images, the graphics processing unit (GPU) writes to the framebuffer while the screen reads this data to display it. If reading and writing overlap, this causes a phenomenon called "tearing", where incomplete parts of the image are displayed, creating an unpleasant visual experience.

**Double Stamp**
Double buffering solves this problem by using two framebuffers: one for the current display and another for the current rendering. Once rendering is complete, the buffers are swapped, ensuring that the full image is displayed without tearing.

**Theatrical illustration**
To explain this technique, an analogy with a play is used: two scenes are prepared. While one scene is played out (lit), the other is prepared in the dark. When the scene changes, the lights instantly switch from one scene to the next, providing a seamless transition for the audience.

**When to use it**
Double buffering is necessary when:
- The system state is changed incrementally.
- This status can be read in the middle of changes.
- It is crucial to present the final state without showing the intermediate steps.
- It is important to be able to read the status without waiting for the modifications to complete.

**Points to consider**
- **Swap time**: The swap process must be fast and atomic so as not to block access to the state.
- **Additional memory**: Double buffering requires double the memory, which can be restrictive on devices with limited resources.
