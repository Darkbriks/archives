# 1. Instanciez un objet de la classe Graphe, et utilisez la méthode afficher_carte pour vérifier que tout marche bien.

```python
import graphelib as gl

g = gl.Graphe()
g.afficher_carte()
```

# 2. Implémentez Dijkstra.

```python
depart = g.dic_noeud[62]
arrivee = g.dic_noeud[74]
predecesseurs, distances, nb_noeuds_dijkstra = g.dijkstra(depart, arrivee)
```

# 3. Ajoutez une fonction permettant de transformer ce que renvoie Dijkstra en le plus court chemin.

```python
plus_court_chemin = gl.plus_court_chemin(predecesseurs, depart, arrivee)
```

# 4. Affichez le résultat sur la carte.

```python
carte = gl.Carte(g)
carte.afficher_chemin(plus_court_chemin)
```

# 5. De même avec $A^*$.

```python
predecesseurs, distances, nb_noeuds_astar = g.astar(depart, arrivee)
plus_court_chemin = gl.plus_court_chemin(predecesseurs, depart, arrivee)
carte = gl.Carte(g)
carte.afficher_chemin(plus_court_chemin)
```

# 6. Ajoutez des compteurs pour comparer le nombre de noeuds visités par les 2 algorithmes.

```python
print(f"Nombre de noeuds visités par Dijkstra: {nb_noeuds_dijkstra}")
print(f"Nombre de noeuds visités par A*: {nb_noeuds_astar}")
```
