# Introduction
## What is a telescope ?
A telescope is an optical instrument that gathers and focuses light to create a magnified image of an object. The image can be viewed through an eyepiece, which acts like a magnifying glass, or by using a camera to record the image onto a photographic plate or CCD (Charge-Coupled Device). Telescopes are used to observe distant objects, such as planets, stars, and galaxies. They are also employed to collect light that is too faint for the naked eye to see, such as nebulae and
galaxies.

In reality, a telescope consists of two main parts: the tube and the mount. The tube is the optical component, and the mount is the mechanical part.

### The tube
The tube is the optical component of the telescope. It can be composed of lenses, mirrors, or a combination of both. The tube is one of the most crucial parts of the telescope since it determines the quality of the image.

The first telescopes were refractors, which used lenses to focus the light. Galileo Galilei was the first to use a telescope to observe the sky. He constructed a telescope with 30x magnification. Initially, refractors were made with a single lens, but it was soon discovered that the image was distorted. To address this issue, a second lens was added, giving birth to the first achromatic refractor. The achromatic refractor remains the least expensive type of telescope.

Later, Newton invented the reflecting telescope, which utilizes mirrors instead of lenses. The first reflecting telescope featured a 6-inch mirror crafted by Newton himself. The reflecting telescope continues to be the most popular type due to its superior diameter-to-price ratio.

### The mount
In astronomy, mounts play a crucial role in the operation of telescopes. They constitute the mechanical part that supports the optical tube, allowing its movement to track celestial objects in motion. Mounts are designed to provide precise stability, minimizing vibrations and enabling smooth and accurate movements. There are different types of mounts, including equatorial mounts that follow the Earth's rotation, and azimuthal mounts that move the telescope horizontally and vertically. The choice of the mount significantly impacts the ease of observation and the quality of images captured by the telescope, making this component a crucial consideration for amateur and professional astronomers.

### The eyepiece
Eyepieces play a fundamental role in astronomical observation, allowing astronomers to appreciate the details and beauty of celestial objects. As essential optical elements of telescopes, eyepieces are positioned at the end of the optical tube to magnify the image formed by the objective lens or main mirror. They determine the magnification power and field of view, influencing the quality of observation. Astronomers choose their eyepieces based on the specific nature of celestial objects they want to study, the clarity of the sky, and observing conditions. With a variety of eyepieces available, ranging from low-magnification models for panoramic views to high-magnification eyepieces for finer details, the choice of eyepiece is an important consideration to optimize the astronomical observation experience.

## A little bit of math
### Definition
The focal length $f$ of a lens is the distance between the lens and the focal point. The focal point is the point where the light rays converge after passing through the lens. The focal length is measured in millimeters (mm).
The diameter $D$ of an optical instrument is the diameter of the objective lens or mirror. The diameter is measured in millimeters (mm). The diameter is also called the aperture.
The focal ratio $f/D$ of an optical instrument is the ratio of the focal length $f$ to the diameter $D$ of the objective lens or mirror. The focal ratio is also called the f-number or f-stop. The focal ratio is a dimensionless number. It determines the brightness of the image. The lower the focal ratio, the brighter the image.

### Magnification
The magnifying power $M$ of a telescope is the measure of the capability of an optical instrument to magnify the apparent size of an object. It is defined as the ratio of the focal length of the objective lens $f_{obj}$ to the focal length of the eyepiece $f_{eye}$:
$$M = \frac{f_{obj}}{f_{eye}}$$
For example, a telescope with a 1000 mm focal length objective lens and a 5mm focal length eyepiece will have a magnification of 200x.
Warning: the magnification is not the most important parameter of a telescope. Indeed, the magnification is limited by the diameter of the objective lens or mirror. The maximum magnification is given by the construction type of the telescope, the quality of the optics, and the atmospheric conditions. In a future section, we will see the maximum theoretical magnification for each type of telescope. If the magnification is too high, the image will be blurred and will not be usable.

### Exit pupil
The exit pupil of a telescope is the apparent size of the focused light beam as it exits the eyepiece. It is calculated by dividing the diameter of the objective lens or mirror by the magnification:
$$\text{Exit Pupil} = \frac{D}{M}$$
The exit pupil is important because it determines how much of the telescope's light is entering your eye. For optimal viewing, the exit pupil should match the size of your eye's pupil, which is typically around 7mm under dark conditions.

### Field of view
The field of view (FOV) is the amount of the sky that is visible through the eyepiece. It is expressed in degrees and can be calculated using the following formula:
$$\text{FOV} = \frac{Apparent Field of View}{Magnification}$$
The apparent field of view of an eyepiece is a fixed value specified by the manufacturer. It is typically between 40° and 100°. The field of view of a telescope is determined by the focal length of the eyepiece and the magnification.

## Tube
### Reflecting telescopes
A reflecting telescope uses a combination of mirrors to focus light. It is the most popular type of telescope due to its superior diameter-to-price ratio. Reflecting telescopes are available in three main designs: Newtonian, Cassegrain, and Ritchey-Chretien.

#### Newtonian
The Newtonian telescope, designed by Sir Isaac Newton in the 17th century, feature relatively simple optical construction. A concave parabolic mirror, called a primary mirror, which collects incident light. This light is then reflected towards a secondary planar mirror inclined at 45 degrees, located near the opening of the tube, and redirected towards an eyepiece placed on the side of the tube. This design helps eliminate chromatic aberration, providing a sharper, more contrasty image than some other types of telescopes. However, it causes the appearance of "coma", an optical abberation which distorts the image at the edge of the field.

##### Pros
- Simple optical construction
- Eliminates chromatic aberration
- Sharper, more contrasty image
- Low cost
- Easy to use
- Good for deep-sky observation
- Good for astrophotography
- Good for planetary observation (for the biggest astronomical objects, like the Moon, Jupiter, Saturn, Venus...)

##### Cons
- Coma
- Very encombrant (a 1000mm focal length telescope is about 1 meter long)
- Heavy (a 1000mm focal length telescope is about 10kg)

#### Cassegrain
The Cassegrain telescope is another type of reflecting telescope. It uses a combination of a concave primary mirror and a convex secondary mirror to fold the light path, allowing for a more compact design. The focused light is directed through a hole in the primary mirror to the eyepiece or camera at the back.

##### Pros
- Compact design
- Very good for planetary observation due to its long focal length
- Good for astrophotography

##### Cons
- Expensive
- Coma
- Heavy
- Difficult to use
- Not good for deep-sky observation due to its big $f/D$ ratio

#### Ritchey-Chretien
The Ritchey-Chretien telescope is a specialized Cassegrain telescope designed to eliminate coma and spherical aberration. It uses hyperbolic primary and secondary mirrors to achieve a wide field of view and a flat image plane. It is popular among professional astronomers for its high image quality.

##### Pros
- Compact design
- Very good for planetary astrophotography due to its long focal length
- No coma
- No spherical aberration
- Wide field of view
- Flat image plane

##### Cons
- Expensive
- Heavy
- Difficult to use
- Not designed for visual observation

### Catadioptric telescopes
A catadioptric telescope uses a combination of mirrors and lenses to focus light. It is a compact design that offers a wide field of view and a flat image plane. Catadioptric telescopes are available in two main variants: Schmidt-Cassegrain and Maksutov-Cassegrain, which as their name suggests, are based on the Cassegrain design.

#### Schmidt-Cassegrain
The Schmidt-Cassegrain telescope is a compact design that combines a spherical primary mirror with a corrector plate and a spherical secondary mirror. The corrector plate is a thin lens that corrects the spherical aberration caused by the primary mirror. The focused light is directed through a hole in the primary mirror to the eyepiece or camera at the back. For certain models, the secondary mirror can be replaced by a camera with a corrector, allowing for a less magnified image.

##### Pros
- Compact design
- Wide field of view
- Flat image plane
- Good for planetary observation
- Good for astrophotography

##### Cons
- Expensive
- Heavy
- Difficult to use
- Not good for deep-sky observation due to its big $f/D$ ratio

#### Maksutov-Cassegrain
The Maksutov-Cassegrain telescope is a compact design that combines a spherical primary mirror with a corrector plate and a spherical secondary mirror. The corrector plate is a thin lens that corrects the spherical aberration caused by the primary mirror. The focused light is directed through a hole in the primary mirror to the eyepiece or camera at the back.

##### Pros
- Compact design
- Wide field of view
- Flat image plane
- Good for planetary observation
- Good for astrophotography
- Less expensive than Schmidt-Cassegrain

##### Cons
- Heavy
- Difficult to use
- Not good for deep-sky observation due to its big $f/D$ ratio

### Refracting telescopes
A refracting telescope uses a combination of lenses to focus light. It is the oldest type of telescope, and it is still used today. Refracting telescopes are available in three main categories, depending on the type of lenses used, and the number of groups of lenses: achromatic refractor, apochromatic refractor, and semi-apo refractor. A fourth category, the astrograph, is a specialized refractor designed for astrophotography, and it is not suitable for visual observation.

#### Achromatic refractor
The achromatic refractor is the simplest type of refracting telescope. It uses two lenses to focus light: a convex objective lens and a concave eyepiece lens. The objective lens collects light and focuses it to a point, called the focal point. The eyepiece lens magnifies the image formed by the objective lens. The focused light is directed through the eyepiece to the observer's eye. The achromatic refractor is the least expensive type of telescope, but it suffers from chromatic aberration, which causes the appearance of color fringes around the image. The chromatic aberration is caused by the fact that the lenses refract light of different wavelengths by different angles, causing the light to focus at different points. It's the same phenomenon that causes a rainbow to appear when light passes through a prism.

##### Pros
- Simple optical construction
- Low cost
- Very easy to use
- Good for planetary observation
- Possible to use for deep-sky observation (depending on the $f/D$ ratio)

##### Cons
- Chromatic aberration
- Very bad for astrophotography
- Heavy (a lens is heavier than a mirror)
- Not possible to create a lens with a diameter greater than $\approx 150mm$

#### Apochromatic refractor
The apochromatic refractor is a specialized refracting telescope designed to eliminate chromatic aberration. It uses three or more group of lenses to focus light. The first group of lenses is a convex objective lens, which collects light and focuses it to a point, called the focal point. The other groups of lenses are concave lenses, which correct the chromatic aberration caused by the objective lens. The focused light is directed through the eyepiece to the observer's eye. The apochromatic refractor one of the most expensive type of telescope, but it offers a very high image quality.

##### Pros
- Very high image quality
- No chromatic aberration
- Good for planetary observation
- Very good for astrophotography
- Possible to use for deep-sky observation (depending on the $f/D$ ratio)
- Very high contrast
- Very sharp image
- Easy to use

##### Cons
- Very expensive
- Heavy (a lens is heavier than a mirror, and there are a minimum of 3x2 lenses)
- Not possible to create a lens with a diameter greater than $\approx 150mm$

#### Semi-Apo refractor
The semi-Apo refractor is an intermediate type of refracting telescope. It uses two group of lenses to reduce chromatic aberration. The first group of lenses is a convex objective lens, which collects light and focuses it to a point, called the focal point. The second group of lenses is a concave lens, which corrects the chromatic aberration caused by the objective lens. The focused light is directed through the eyepiece to the observer's eye. The semi-Apo refractor is less expensive than the apochromatic refractor, but it offers a lower image quality. For most models, it's possible to upgrade the tube with a corrector lens placed between the returns angled and the eyepiece, to reduce the chromatic aberration.

##### Pros
- High image quality
- Very low chromatic aberration (can be reduced with a corrector lens)
- Good for planetary observation
- Good for astrophotography
- Possible to use for deep-sky observation (depending on the $f/D$ ratio)
- High contrast
- Sharp image
- Easy to use
- Less expensive than apochromatic refractor

##### Cons
- Expensive
- Heavy (a lens is heavier than a mirror, and there are a minimum of 2x2 lenses)
- Not possible to create a lens with a diameter greater than $\approx 150mm$
- More chromatic aberration than apochromatic refractor

#### Astrograph
The astrograph is a specialized refracting telescope designed for astrophotography. It uses a very high-quality lens, with a very low chromatic aberration, to focus light. The focused light is directed through the camera. The astrograph is not suitable for visual observation.

##### Pros
- The best image quality
- No visible chromatic aberration
- Specialized for astrophotography

##### Cons
- Very expensive
- Heavy (a lens is heavier than a mirror)
- Not possible to create a lens with a diameter greater than $\approx 150mm$
- Not suitable for visual observation

## Mount
The mount is the mechanical part that supports the optical tube, allowing its movement to track celestial objects in motion. Mounts are designed to provide precise stability, minimizing vibrations and enabling smooth and accurate movements. There are different types of mounts, including equatorial mounts that follow the Earth's rotation, and azimuthal mounts that move the telescope horizontally and vertically. The choice of the mount significantly impacts the ease of observation and the quality of images captured by the telescope, making this component a crucial consideration for amateur and professional astronomers.

### Alt-Azimuth
The alt-azimuth mount is the simplest type of mount. It is composed of two axes: the altitude axis and the azimuth axis. The altitude axis allows the telescope to move vertically, and the azimuth axis allows the telescope to move horizontally. The alt-azimuth mount is the most common type of mount for amateur astronomers. It is easy to use and inexpensive. However, it is not suitable for astrophotography because it does not follow the Earth's rotation with only one axis, and causes a rotation of the image during long exposures.

#### Pros
- Easy to use
- Inexpensive
- Good for visual observation
- No need to polar align
- Can be motorized with a GoTo system (the telescope will automatically move to the selected object, and follow it)

#### Cons
- Not suitable for astrophotography

### Equatorial
The equatorial mount is a specialized mount designed to follow the Earth's rotation. It is composed of four axes: the right ascension axis, the declination axis, the altitude axis, and the azimuth axis. The altitude and azimuth axes is used to point the telescope to the celestial pole. The right ascension axis is used to follow the Earth's rotation. The declination axis is used to point the telescope to the celestial equator. The equatorial mount is the most common type of mount for astrophotography. It is more expensive than the alt-azimuth mount, but it is the only type of mount that can follow the Earth's rotation without causing a rotation of the image during long exposures.

#### Pros
- Good for astrophotography
- Can follow the Earth's rotation
- Can be motorized with a GoTo system (the telescope will automatically move to the selected object, and follow it)
- Can be used for visual observation

#### Cons
- Expensive
- Difficult to use
- Need to polar align
- Need to balance the telescope with counterweights
- More heavy than alt-azimuth mount for the same util capacity

*Note: during last years, a new type of equatorial mount has been developed by ZWO and iOptron, the harmonic drive mount. This type of mount is very expensive, but it is very light. I will not talk about this type of mount because it is not very common for the moment, and his fonctionnement is very complex (it uses a harmonic drive, which is a very complex mechanical system, also used for Curiosity rover on Mars, for example).*

### Dobsonian
The Dobsonian mount is a relatively new type of mount, invented by John Dobson in the 1960s. It is a variation of the alt-azimuth mount, with a very simple design. It is composed of two axes: the altitude axis and the azimuth axis. The altitude axis allows the telescope to move vertically, and the azimuth axis allows the telescope to move horizontally. The Dobsonian mount is the most common type of mount for amateur astronomers. It is easy to use and it's the least expensive type of mount. However, it is not suitable for astrophotography because it suffers of the same problem as the alt-azimuth mount.

#### Pros
- Easy to use
- Least expensive type of mount
- Good for visual observation
- No need to polar align
- Can be motorized with a GoTo system (the telescope will automatically move to the selected object, and follow it)
- Very light

#### Cons
- Not suitable for astrophotography

## Other accessories
To practice astronomy, you need a telescope, but you also need other accessories. In this section, I will talk about the most common accessories for visual observation and astrophotography.

### Finder
The finder is a small telescope that is mounted on the main telescope. It is used to point the main telescope to the selected object. It is composed of a small objective lens, and an eyepiece. The finder is very useful because it has a very wide field of view, allowing to easily find the selected object. There are two types of finder: the optical finder and the red dot finder. The optical finder is a small refractor telescope, and the red dot finder is a small red dot that is projected on a transparent screen. The red dot finder is less expensive than the optical finder, and it is easier to use, but less precise. A very large number of telescopes are sold with a red dot finder, but it is possible to buy an other finder separately.

### Focuser and microfocuser
The focuser is the part of the telescope that holds the eyepiece. It allows to move the eyepiece to focus the image. The focuser can be simple or Crayford. The Crayford focuser is a more precise type of focuser, because it uses a rack and pinion system allowing a very precise movement of the eyepiece, and a dual speed system allowing a very precise focusing. The focuser can be motorized with a microfocuser. The microfocuser is a small motor that allows to move the eyepiece very precisely. It is very useful for astrophotography, because it allows to focus the image without touching the telescope, which can cause vibrations.

### Visual accessories
#### Eyepieces
The eyepiece is the part of the telescope that is placed at the end of the optical tube. It is composed of a lens, and a barrel. The eyepiece is used to magnify the image formed by the objective lens or mirror. There are an important number of eyepieces configurations, like the Plössl, the Kellner, the Orthoscopic... The eyepiece is one of the most important part of the telescope in visual, because it determines the magnification, the field of view, and the image quality. The eyepiece is necessary for visual observation, and always sold with the telescope. However, it is possible to buy another eyepiece separately, to have a better image quality, or a different magnification or field of view. Generally, you must have at least two or three eyepieces with different focal lengths, to have a wide range of magnifications and fields of view.

#### Barlow
The Barlow is an accessory that is placed between the eyepiece and the focuser. It is composed of a divergent lens, and a barrel. The Barlow is used to increase the magnification of the eyepiece. It is very useful because it allows to have a very large range of magnifications with only two or three eyepieces. The Barlow is not necessary, but it is very useful for visual observation, but not for astrophotography, because it reduces the image quality. The most commons Barlow are 2x, 3x, and 5x (the magnification is multiplied by the Barlow factor).

### Astrophotography accessories
#### Camera
The camera is the most important part of the astrophotography setup. It is used to capture the image of the selected object. There are two major types of cameras: the DSLR (Digital Single Lens Reflex) camera, and the CCD (Charge-Coupled Device) camera. The DSLR camera is a "classic" camera, like the Canon EOS 600D, or the Nikon D5300. The DSLR camera is very easy to use, and it is good for astrophotography. The CCD camera is a specialized camera, like the ZWO ASI 1600MM Pro, or the QHYCCD QHY183M. The CCD camera is more difficult to use than the DSLR camera, but it is more sensitive, and it has a very low noise thanks to the cooling system. The CCD camera is the best type of camera for astrophotography, but it is very expensive.

#### Focal reducer
The focal reducer is an accessory that is placed between the focuser and the camera. It is composed of a convergent lens, and a barrel. The focal reducer is used to reduce the focal length of the telescope, and therefore the focal ratio. It is very useful for astrophotography, because it allows to have a wider field of view, and a brighter image. The focal reducer is not necessary, but it is very useful for astrophotography. The most commons focal reducers are 0.5x, 0.75x, and 0.8x (the focal length is multiplied by the focal reducer factor).

#### Field flattener
The field flattener is an accessory that is placed between the focuser and the camera. It is composed of a convergent lens, and a barrel. The field flattener is used to flatten the image, and therefore to reduce the curvature of the image. It is very useful for astrophotography, because it allows to have a flat image, and therefore a better image quality. The field flattener is not necessary, but it is very useful for astrophotography. The most commons field flatteners are 0.8x, and 1x (the focal length is multiplied by the field flattener factor).

#### Autoguider
The autoguider is a comlpex system composed of a camera, a guide scope, and a computer. The autoguider is used to correct the tracking of the mount. It is very useful for astrophotography, because it allows to have a very precise tracking, and therefore to have a better image quality. The autoguider is not necessary, but it is very useful for astrophotography when the exposure time is long (more than 30 seconds).

#### Filters and filter wheel
The filters are accessories that are placed between the focuser and the camera. The filters are used to filter the light of the selected object. There are many types of filters, like the Luminance filter, the Red filter, the Green filter, the Blue filter, the Hydrogen Alpha filter, the Oxygen III filter, the Sulfur II filter... The filters are very useful for monochrome cameras, because they allow to have a color image. Filters also can used to reduce the light pollution, or to increase the contrast of the image. The filter wheel is an accessory that is placed between the focuser and the camera. It is composed of a wheel, and a barrel. The filter wheel is used to change the filter more easily.

#### Coma corrector
The coma corrector is an accessory used for Newtonian telescopes. It is placed between the focuser and the camera. It is composed of a convergent lens, and a barrel. The coma corrector is used to correct the coma, an optical aberration that causes the appearance of a comet-like shape on the image. The coma corrector is not necessary, but it is very useful for increasing the image quality of Newtonian telescopes.

#### Mask
##### Bahtinov mask
The Bahtinov mask is an accessory used for focusing the image. It is placed at the end of the optical tube. It is composed of a mask with slots. The Bahtinov mask creates a diffraction pattern, which is used to easily focus the image.

##### Hartmann mask
The Hartmann mask is another accessory used for focusing the image, with a different pattern than the Bahtinov mask, composed of a mask with holes.

##### Chanfrein mask
The Chanfrein mask is positioned in the primary mirror of a reflecting telescope. It's used to reduce the diffraction spikes caused by the spider of the telescope.

- **Exit pupil**: the apparent size of the focused light beam as it exits the eyepiece. It is calculated by dividing the diameter of the objective lens or mirror by the magnification.
- **Eyepiece**: the part of the telescope that is placed at the end of the optical tube. It is composed of a lens, and a barrel. The eyepiece is used to magnify the image formed by the objective lens or mirror.
    - <img src="Images/oculaires.jpg" alt="Eyepiece" width="300"/>
    - Shema of some eyepieces:
    - <img src="Images/oculaires-schema.jpg" alt="Eyepiece schema" width="300"/>
- **Field flattener**: an accessory that is placed between the focuser and the camera. It is composed of a convergent lens, and a barrel. The field flattener is used to flatten the image, and therefore to reduce the curvature of the image.
    - <img src="Images/field-flattener.jpg" alt="Field flattener" width="300"/>
    - Effect of the field flattener:
    - <img src="Images/field-flattener-exemple.jpg" alt="Field flattener effect" width="300"/>
- **Field of view**: the amount of the sky that is visible through the eyepiece. It is expressed in degrees.
- **Finder**: a small telescope that is mounted on the main telescope. It is used to point the main telescope to the selected object.
    - <img src="Images/finder.jpg" alt="Finder" width="300"/>
- **Focal length**: the distance between the lens and the focal point. The focal length is measured in millimeters (mm).
- **Focal ratio**: the ratio of the focal length to the diameter of the objective lens or mirror. The focal ratio is also called the f-number or f-stop. The focal ratio is a dimensionless number. It determines the brightness of the image. The lower the focal ratio, the brighter the image.
- **Focal reducer**: an accessory that is placed between the focuser and the camera. It is composed of a convergent lens, and a barrel. The focal reducer is used to reduce the focal length of the telescope, and therefore the focal ratio.
    - <img src="Images/reducteur-de-focale.jpg" alt="Focal reducer" width="300"/>
- **Focuser**: the part of the telescope that holds the eyepiece. It allows to move the eyepiece to focus the image.
    - <img src="Images/dual-speed_focuser.jpg" alt="Focuser" width="300"/>
- **Harmonic drive**: a type of equatorial mount. This type of mount is very expensive, but it is very light.
    - <img src="Images/zwo_am5.jpg" alt="Harmonic drive" width="300"/>
- **Hartmann mask**: an accessory used for focusing the image, with a different pattern than the Bahtinov mask, composed of a mask with holes.
    - <img src="Images/hartmann.jpg" alt="Hartmann mask" width="300"/>
- **Image quality**: the quality of the image formed by the objective lens or mirror. It is determined by the quality of the optics, and the atmospheric conditions.
- **Magnification**: the measure of the capability of an optical instrument to magnify the apparent size of an object. It is defined as the ratio of the focal length of the objective lens to the focal length of the eyepiece.
- **Maksutov-Cassegrain**: a type of catadioptric telescope that uses a combination of a spherical primary mirror with a corrector plate and a spherical secondary mirror. The corrector plate is a thin lens that corrects the spherical aberration caused by the primary mirror.
    - <img src="Images/maksutov-cassegrain.jpg" alt="Maksutov-Cassegrain" width="300"/>
    - Shema of a Maksutov-Cassegrain telescope:
    - <img src="Images/maksutov-cassegrain-explaination.png" alt="Maksutov-Cassegrain diagram" width="300"/>
- **Mask**: an accessory used for focusing the image. It is placed at the end of the optical tube. It is composed of a mask with slots or holes. The mask creates a diffraction pattern, which is used to easily focus the image.
- **Microfocuser**: a small motor that allows to move the eyepiece very precisely. It is very useful for astrophotography, because it allows to focus the image without touching the telescope, which can cause vibrations.
- **Mount**: the mechanical part that supports the optical tube, allowing its movement to track celestial objects in motion.
- **Newtonian**: a type of reflecting telescope that uses a concave parabolic mirror, called a primary mirror, which collects incident light. This light is then reflected towards a secondary planar mirror inclined at 45 degrees, located near the opening of the tube, and redirected towards an eyepiece placed on the side of the tube.
    - <img src="Images/newton.jpg" alt="Newtonian" width="300"/>
    - Shema of a Newtonian telescope:
    - <img src="Images/newtonianexplaination.png" alt="Newtonian diagram" width="300"/>
- **Objective lens**: the lens that collects light and focuses it to a point, called the focal point.
- **Optical aberration**: an optical defect that causes the appearance of a defect on the image. There are many types of optical aberrations, like the chromatic aberration, the coma, the spherical aberration...
- **Optical tube**: the optical component of the telescope. It can be composed of lenses, mirrors, or a combination of both.
- **Planetary observation**: the observation of the planets of the solar system. This type of observation requires a high focal, over $\approx 1000mm$, but, isn't necessary to have a high diameter, because the planets are very bright.
- **Primary mirror**: the mirror that collects incident light. This light is then reflected towards a secondary mirror.
- **Red dot finder**: a small red dot that is projected on a transparent screen. The red dot finder is less expensive than the optical finder, and it is easier to use, but less precise. It's very similar to the reflex sight used in airsoft.
    - <img src="Images/red-pointer.jpg" alt="Red dot finder" width="300"/>
- **Reflector**: a type of telescope that uses a combination of mirrors to focus light.
- **Refractor**: a type of telescope that uses a combination of lenses to focus light.
- **Right ascension axis**: the axis of the equatorial mount that is used to follow the Earth's rotation.
- **Ritchey-Chretien**: a type of catadioptric telescope that uses hyperbolic primary and secondary mirrors to achieve a wide field of view and a flat image plane.
    - <img src="Images/Ritchey-Chretien.jpg" alt="Ritchey-Chretien" width="300"/>
    - Shema of a Ritchey-Chretien telescope:
    - <img src="Images/Ritchey-Chrétien-explaination.png" alt="Ritchey-Chretien diagram" width="300"/>
- **Secondary mirror**: the mirror that receives the light reflected by the primary mirror, and reflects it towards the eyepiece.
- **Semi-Apo refractor**: a refracting telescope that uses two group of lenses to reduce chromatic aberration. The first group of lenses is a convex objective lens, which collects light and focuses it to a point, called the focal point. The second group of lenses is a concave lens, which corrects the chromatic aberration caused by the objective lens.
    - <img src="Images/80ed.jpg" alt="Semi-Apo refractor" width="300"/>
- **Spherical aberration**: an optical aberration that causes the appearance of a blur on the image. It is caused by the fact that the light rays are not parallel, but convergent.
    - <img src="Images/Spherical-Aberration.png" alt="Spherical aberration" width="300"/>
- **Spider**: the structure that holds the secondary mirror. It usually has tree or four arms.
    - <img src="Images/spider.jpg" alt="Spider" width="300"/>
- **Tube**: the optical component of the telescope. It can be composed of lenses, mirrors, or a combination of both.
- **Tube rings**: the rings that hold the optical tube on the mount.
    - <img src="Images/tube_rings.jpg" alt="Tube rings" width="300"/>
- **Visual observation**: the observation of the sky with the eye, without camera.
