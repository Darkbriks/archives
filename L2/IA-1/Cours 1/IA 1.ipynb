## Création et importation de dataframe

Regardons la construction d'une DataFrame à partir de séries.

Chaque série (Series) définit une colone du dataframe.

Une manière de le faire est de définir un dictionnaire qui contient pour clef le nom de la colone et pour valeur le série correspondante.

```python
import numpy as np
import pandas as pd

age = pd.Series([30, 20, 50, 74], index=['Alice', 'Bob', 'Julie', 'Antoine'])
height = pd.Series([150, 170, 182, 168], index=['Alice', 'Marc', 'Antoine', 'Julie'])

stat=pd.DataFrame({'age' : age, 'height' : height})
print(stat)
```

## Contenu d'un dataframe et exploration.

```python
#On peut maintenant accéder aux index des lignes et des colonnes
print(stat.index)
print(stat.columns)
```

```python
a = stat.loc['Alice', 'age']
print(a)
print(type(a))
```

```python
#Sélection d'une partie, ici la colonne age
c=stat.loc[:,'age']
print(c)
print(type(c))
```

```python
#Calcul de sa moyenne
m=c.mean()
print(f"L'âge moyen est de {m:.1f} ans")
```
