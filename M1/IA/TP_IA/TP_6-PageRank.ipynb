```python
import numpy as np

def ComputeTransitionMatrix(filename, lambda_=0.85):
    with open(filename, 'r') as f:
        lines = f.readlines()
    n = int(lines[0].strip().split()[0])
    M = np.zeros((n, n))
    for line in lines[1:]:
        parts = list(map(int, line.strip().split()))
        page, links = parts[0], parts[1:]
        if len(links) == 0:
            M[:, page] = 1.0 / n
        else:
            # lambda_*(Aij / SIGMA j=1..n Ajj) + (1-lambda_)*(1/n)
            for link in links:
                M[link, page] = lambda_ / len(links) + (1 - lambda_) / n
    return M.transpose()
```

```python
P = ComputeTransitionMatrix('page_rank_graph.txt', 1)
print("Transition Matrix M:")
print(P)
```

```python
def PageRankPowerMethod(A_filename, lambda_, epsilon=1e-6):
    P = ComputeTransitionMatrix(A_filename, lambda_)
    l = 0
    pgv = np.ones(P.shape[0]) / P.shape[0]
    while True:
        new_pgv = pgv @ P # @ is matrix multiplication
        delta = np.linalg.norm(new_pgv - pgv, 2)
        l += 1
        pgv = new_pgv
        if delta <= epsilon:
            break
    return pgv
```

```python
r = PageRankPowerMethod('page_rank_graph.txt', 0.85)
print("PageRank Vector r:")
print(r)
```
