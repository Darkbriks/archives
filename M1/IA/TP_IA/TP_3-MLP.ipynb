## 1.1 Définir un classifieur MLP pour appredendre l'operateur AND :

```python
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier

X = [[0,0],[0,1],[1,0],[1,1]]
y = [0,0,0,1]

clf = MLPClassifier(hidden_layer_sizes=(), activation='identity', solver="lbfgs")
clf.fit(X,y)

for x in X:
    print(f"AND({x}) ; expected={y[X.index(x)]} ; predicted={clf.predict([x])[0]}")
```

# Classification d'images

```python
from  sklearn.datasets import  load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

dataset = load_digits()
X = dataset.data
y = dataset.target
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.1, random_state=0)

best_activation = ""
best_solver = ""
best_hidden_layer_size = ()
best_accuracy = 0
for activation in ["identity", "logistic", "tanh", "relu"]:
    for solver in ["lbfgs", "sgd", "adam"]:
        for hidden_layer_size in [(), (50,), (100,), (100, 100)]:
            clf = MLPClassifier(hidden_layer_sizes=hidden_layer_size, activation=activation, solver=solver, max_iter=1000)
            clf.fit(train_X, train_y)

            test_y_pred = clf.predict(test_X)
            acc = accuracy_score(test_y, test_y_pred)

            cm = confusion_matrix(test_y, test_y_pred)

            #disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=dataset.target_names)
            #disp.plot(cmap=plt.cm.Blues)
            #plt.suptitle("Matrice de confusion Digits - {} - {} - {}".format(act, solver, hidden_layer_size))
            #plt.title("Accuracy {}".format(acc))
            #plt.show()
            print(f"{activation} ; {solver} ; {acc} ; {hidden_layer_size}")

            if acc > best_accuracy:
                best_accuracy = acc
                best_solver = solver
                best_activation = activation
                best_hidden_layer_size = hidden_layer_size

print(f"\nBest results : accuracy {best_accuracy} ; solveur {best_solver} ; activation {best_activation} ; hidden_layer_size {hidden_layer_size}")
```
